## Step 2: Connect to Google Gemini

To access Gemini models, we need to authenticate using an API key.

Think of the API key as a secure password that allows our application to communicate with Google's AI services.

```python
from google import genai

client = genai.Client(
    api_key="GEMINI_API_KEY"
)

print("Connected Successfully!")
```

```python
models = client.models.list()

for model in models:
    print(model.name)
    break
```

## Step 3: Collect User Inputs

Now we'll collect the details from the user.

These inputs will help our AI generate personalized social media content based on:
- Topic
- Platform
- Tone
- Target Audience
- Content Type

```python
topic = input("Enter the topic: ")

platform = input("Choose platform (LinkedIn/Instagram/X): ")

tone = input("Choose tone (Professional/Friendly/Funny): ")

audience = input("Target audience: ")

content_type = input("Content type (Post/Caption/Carousel): ")
```

## Step 4: Create the AI Prompt

This is the most important step.

Here, we'll combine all the user inputs into a well-structured prompt.

The better the prompt, the better the AI-generated content.

```python
prompt = f"""
You are an expert Social Media Content Creator.

Generate high-quality content based on the following details.

Topic: {topic}
Platform: {platform}
Tone: {tone}
Target Audience: {audience}
Content Type: {content_type}

Return the output in the following format only:

## 📢 Social Media Post
(Write one engaging post)

## 🏷️ Hashtags
(Generate exactly 5 trending hashtags)

## 📣 Call-To-Action
(Write one engaging CTA)

## 💡 Next Content Idea
(Suggest one related content idea)

Keep the language engaging, concise, and platform-specific.
"""
```

## Step 5: Generate Content using Google Gemini

Now that our prompt is ready, we'll send it to the Gemini AI model.

The model will understand our instructions and generate high-quality social media content based on the user's inputs.

```python
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)
```

## Step 6: Display the Generated Content

Let's display the AI-generated social media content.

```python
print(response.text)
```

## Step 7: Improve the Content

One of the biggest advantages of Generative AI is that we don't have to start from scratch.

We can ask the AI to improve, rewrite, shorten, or change the tone of previously generated content simply by giving it another prompt.

```python
improvement_prompt = f"""
Rewrite the following social media content.

Content:

{response.text}

Make it:

- More engaging
- More professional
- More concise
- Add relevant emojis
- Keep it suitable for {platform}
"""

improved_response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=improvement_prompt
)

print(improved_response.text)
```

## Step 9: Generate an AI Image Prompt

Great content deserves great visuals!

Let's ask AI to generate a professional image prompt that can be used in AI image generation tools to create social media graphics.

```python
image_prompt = f"""
You are an expert AI Image Prompt Engineer.

Generate one highly detailed image prompt for the following topic.

Topic: {topic}

Target Audience: {audience}

The image should look professional, modern, attractive, and suitable for posting on {platform}.

Only generate the image prompt.
"""

image_response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=image_prompt
)

print(image_response.text)
```

## Step 12: Analyze the Generated Image using Gemini

Generative AI is not limited to text generation.

Gemini is a multimodal AI model, which means it can understand both text and images.

Let's upload our generated image and ask Gemini to analyze whether it matches the social media content we created.

```python
analysis_prompt = f"""
You are an expert social media marketing consultant.

Analyze this image based on the following details.

Topic: {topic}

Platform: {platform}

Target Audience: {audience}

Please answer:

1. Does this image match the topic?
2. Is it suitable for {platform}?
3. What emotions does it convey?
4. What could be improved?
5. Give the image a rating out of 10.
"""
```

```python
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=[
        analysis_prompt,
        image
    ]
)

print(response.text)
```
